```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**VALORES NULOS**

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Citas_Digital_Filtrado.csv')
#Los nombres de columna traen espacios extra (ej. 'BDC ', 'Nombre Cliente '), los limpiamos
data.columns = data.columns.str.strip()
data.head(5)
```

```python
#Revisamos tipos de dato y dimensiones del dataset
data.info()
```

```python
#Identificar valores nulos por columna
valores_nulos = data.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe (total)
valores_nulos = data.isnull().sum().sum()
valores_nulos
```

```python
#Realizamos una copia del dataframe
data3 = data.copy()
```

Fecha

Con 89% de valores nulos (solo 72 de 673 registros tienen una fecha real), imputar esta columna con la media, la moda o un texto fijo no aportaría información real, ya que casi todo su contenido sería inventado o repetido. Siguiendo la misma lógica de eliminación de columnas, la mejor decisión aquí es descartar la columna completa en vez de simular un dato que no existe.

```python
#Primer método: eliminación de columna innecesaria
#axis = 1 para columnas (misma lógica que en el notebook original)
data3 = data3.drop("Fecha", axis=1)
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

BDC

Se usa ffill, asumiendo que registros consecutivos en el archivo suelen pertenecer al mismo BDC, por lo que replicar el valor previo es razonable.

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante "forward fill" ("ffill")
data3["BDC"] = data3["BDC"].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Nombre Cliente

Aquí se usa sustitución por un valor de texto específico, ya que no existe un "nombre promedio" que tenga sentido imputar.

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto
data3["Nombre Cliente"] = data3["Nombre Cliente"].fillna("Cliente sin nombre registrado")
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Estatus de Lead

Se usa relleno hacia atrás (`bfill`), tomando el estatus del registro inmediato siguiente como referencia de continuidad del flujo de leads.

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás "backward fill" ("bfill")
data3["Estatus de Lead"] = data3["Estatus de Lead"].bfill()
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Potencial de compra, probabilidad 0–1

La columna llegó como texto porque incluye capturas erróneas. Primero se convierte a numérico (los valores no numéricos quedan como nulo) y después se aplica sustitución con la mediana, más robusta que la media ante esos valores atípicos de captura.

```python
#Quinto método de sustitución de valores nulos
#Convertimos a numérico (valores mal capturados se vuelven NaN) y sustituimos con la mediana
data3["Potencial de compra"] = pd.to_numeric(data3["Potencial de compra"], errors="coerce")
data3["Potencial de compra"] = data3["Potencial de compra"].fillna(round(data3["Potencial de compra"].median(), 2))
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

PDM — numérica

Solo 1 valor nulo y sin outliers extremos, por lo que se usa sustitución con el promedio (media).

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["PDM"] = data3["PDM"].fillna(round(data["PDM"].mean(), 2))
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

SDC

Se usa sustitución con la mediana, priorizando robustez sobre la media.

```python
#Séptimo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data3["SDC"] = data3["SDC"].fillna(round(data["SDC"].median(), 2))
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Venta

Se usa sustitución por un valor numérico específico (0) y si no quedó registrada, se asume que no hubo venta.

```python
#Octavo método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en concreto
data3["Venta"] = data3["Venta"].fillna(0)
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Asesor Asignado — variable categórica

Se usa relleno hacia adelante (`ffill`), igual razonamiento que en BDC, la continuidad entre registros consecutivos.

```python
#Noveno método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante "forward fill" ("ffill")
data3["Asesor Asignado"] = data3["Asesor Asignado"].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos = data3.isnull().sum()
valores_nulos
```

Columnas de texto libre condicional

Estas 5 columnas son comentarios abiertos que solo aplican cuando la respuesta a la pregunta implícita es relevante para ese lead (por ejemplo "¿por qué no ha visitado la agencia?" solo tiene sentido si no la ha visitado). Un nulo aquí probablemente significa que la pregunta no aplicaba al caso, no que falte el dato. Se usa sustitución por un valor de texto específico (No aplica) en las cinco.

```python
#Décimo método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto, en las columnas de texto condicional
columnas_condicionales = [
    "¿Por que no ha visitado la agencia?",
    "¿Por que no hay solicitud de credito?",
    "¿Por que no hubo prueba de manejo ?",
    "¿Por que no hubo proceso wow?",
    "¿Por que se asigno antes de vistar piso?"
]
for columna in columnas_condicionales:
    data3[columna] = data3[columna].fillna("No aplica")
```

```python
#Corroboramos valores nulos en todo el dataframe
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
#Corroboramos valores nulos en todo el dataframe (total)
valores_nulos = data3.isnull().sum().sum()
valores_nulos
```

Método de Eliminación de valores nulos

Como alternativa de referencia (no se usa para el archivo final), podemos ver lo que pasaría si en vez de sustituir se eliminaran directamente todas las filas con al menos un valor nulo. Se parte de data sin la columna Fecha (ya descartada por columna) para que la comparación sea justa frente a data3.

```python
#Realizamos una copia del dataframe, ya sin la columna Fecha
data4 = data.drop("Fecha", axis=1)
#Identificar valores nulos por columna antes de eliminar filas
valores_nulos = data4.isnull().sum()
valores_nulos
```

```python
#Eliminamos filas con nulos
#No se agrega nada para eliminar filas y 1 para eliminar columnas
data4 = data4.dropna()
data4
```

```python
#Identificar valores nulos por columna
valores_nulos = data4.isnull().sum()
valores_nulos
```

El resultado de dropna  queda con 0 filas. Con 13 columnas y varias de ellas con cientos de nulos (hasta 320 en una sola columna), prácticamente todas las filas tienen al menos un valor nulo en alguna columna, así que basta uno solo para descartar la fila completa. Esto confirma que la eliminación no es viable para este archivo  (perdería el 100% de los registros) y refuerza que la combinación de eliminar la columna Fecha + sustituir el resto por columna data3 es lo mejor, ya que se conservan las 673 filas originales con información real en cada campo.
